```python
class Disciplina:
    def __init__(self, nome, professor):
        self.nome = nome
        self.professor = professor
```

```python
class Aluno:
    def __init__(self, nome):
        self.nome = nome
        self.disciplinas = []
        self.notas = {}
```

```python
    def adicionar_nota(self, disciplina, nota):
        if disciplina not in self.disciplinas:
            self.disciplinas.append(disciplina)

        if disciplina.nome not in self.notas:
            self.notas[disciplina.nome] = []

        self.notas[disciplina.nome].append(nota)
```

```python
# Adicionando o método criado acima à classe Aluno de forma dinâmica
Aluno.adicionar_nota = adicionar_nota
```

```python
a = Aluno("João")
d1 = Disciplina(nome="Python", professor="Prof Ana")
d2 = Disciplina(nome="Python", professor="Prof Carlos")
```

```python
a.adicionar_nota(d1, nota=8)
a.adicionar_nota(d2, nota=6)
a.adicionar_nota(d1, nota=10)
```

```python
print(f"Quantidade de disciplinas: {len(a.disciplinas)}")
print(f"Disciplinas com notas lançadas: {len(a.notas)}")
print(f"Notas de Python: {a.notas['Python']}")
```
